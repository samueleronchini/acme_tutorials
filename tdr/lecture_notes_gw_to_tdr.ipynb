## 0. Motivation (4 min)

**Template event.** GW170817 / GRB 170817A: a BNS inspiral at ≈ 40 Mpc seen in gravitational waves, a short GRB ≈ 1.7 s after the merger, and the kilonova AT2017gfo. Mergers involving at least one NS can power emission from radio to very high energies (paper, Sect. 1): short-GRB prompt emission and afterglow; nuclear-decay-powered kilonovae; soft, fast X-ray transients from outside the jet cone; emission from a shock breaking out of the post-merger ejecta; magnetar spin-down emission; plausibly also high-energy neutrinos and cosmic rays. During the first four LVK observing runs the interferometers probed BNS mergers out to ≈ 200–300 Mpc and NSBH mergers out to ≈ 1 Gpc (paper, Sect. 1).

**Practical problem.** When a non-GW trigger arrives (GRB, fast X-ray transient, optical transient, neutrino alert), follow-up resources (deep optical spectroscopy, soft X-ray, radio, VHE γ-ray) are scarce and must be allocated within minutes to days, increasingly so with wide-field facilities such as Einstein Probe and the Vera Rubin Observatory. If no LVK candidate coincident with the trigger has been released on GraceDB, the operative question is: *could the operating interferometers have detected a merger associated with this trigger?* The answer (a) prioritizes follow-up and (b) constrains the physical origin of the transient.

**Existing tools and their gaps.**

- *BNS range* (published in real time for each IFO): a 1.4 + 1.4 M⊙ binary averaged over sky position and inclination. It characterizes the detector and ignores everything the EM trigger reveals.
- *Targeted LVK searches*: PyGRB (coherent matched filtering for CBC signals) and X-Pipeline (unmodelled transients). Sensitive and calibrated, but computationally expensive, with latencies of the order of hours; a targeted GstLAL version is ≈ 50 times faster than its all-sky counterpart (paper, Sect. 1).

**Idea of the paper.** Replace the search by an *expected-sensitivity calculation*: draw simulated signals consistent with the priors set by the EM observation (sky position, inclination, masses), evaluate their SNR in the noise of the IFOs that were observing at the trigger time, and report the distance `D₉₀` within which 90% of them exceed an SNR threshold. The cost is minutes rather than hours.

**Roadmap.** The TDR is assembled from five ingredients: detector response (§1), waveform (§2), noise and SNR statistics (§3), distance scaling (§4), and the logic of targeted searches and exclusion distances (§5). §6 combines them.

## 1. Gravitational-wave basics (8 min)

### 1.1 Linearized gravity and polarizations

In the weak-field regime $g_{\mu\nu} = \eta _{\mu\nu} + h_{\mu\nu}$ with $|h_{\mu\nu}| \ll  1$. In Lorenz gauge the trace-reversed perturbation obeys a wave equation with a source,

$$
\begin{gathered}
\Box \bar{h}_{\mu\nu} = -(16\pi G/c^{4}) T_{\mu\nu}
\end{gathered}
$$

so that h propagates at speed c in vacuum. The transverse-traceless (TT) gauge leaves two independent components; for propagation along z,

$$
\begin{gathered}
h_{\mathrm{xx}} = -h_{\mathrm{yy}} = h_{+}(t - z/c),\quad h_{\mathrm{xy}} = h_{\mathrm{yx}} = h_{\times}(t - z/c),\quad\text{all others zero}
\end{gathered}
$$

Free test masses respond through geodesic deviation, $\ddot{\xi }^{i} = \frac{1}{2} \ddot{h}_{ij}^{\mathrm{TT}} \xi ^{j}$, so that $\delta L/L = \frac{1}{2} h$: h is a *strain*. A ring of test masses is stretched along the x and y axes by h<sub>+</sub> and along the diagonals by h× (a spin-2 field: the two states differ by a 45° rotation). Astrophysical strains are minute, h $\sim$ 10<sup>-21</sup>–10<sup>-23</sup>, i.e. δL $\sim$ 10<sup>-18</sup>–10<sup>-20</sup> m for L = 4 km.

### 1.2 Detector response and antenna patterns

For a Michelson interferometer with orthogonal arms and a GW wavelength much longer than the arms (f ≪ c/L ≈ 75 kHz for L = 4 km) the recorded strain is

$$
\begin{gathered}
h(t) = F_{+}(\theta ,\phi ,\psi ) h_{+}(t) + F_{\times}(\theta ,\phi ,\psi ) h_{\times}(t) \\
F_{+} = \frac{1}{2}(1+\cos^{2}\theta ) cos2\phi  cos2\psi  - \cos\theta  sin2\phi  sin2\psi  \\
F_{\times} = \frac{1}{2}(1+\cos^{2}\theta ) cos2\phi  sin2\psi  + \cos\theta  sin2\phi  cos2\psi 
\end{gathered}
$$

where (θ, φ) is the source direction in the detector frame (polar axis normal to the arm plane, x and y along the arms) and ψ is the polarization angle. Points for the board:

- $F_{+}^{2} + F_{\times}^{2} = \frac{1}{4}(1+\cos^{2}\theta )^{2} \cos^{2}2\phi  + \cos^{2}\theta  \sin^{2}2\phi $ is **independent of ψ**. It equals 1 for a source overhead (θ = 0) and vanishes in four directions of the arm plane, along the arm bisectors (θ = π/2, φ = π/4 + nπ/2).
- Averaged over sky and polarization, $\langle F_{+}^{2}\rangle  = \langle F_{\times}^{2}\rangle  = 1/5$.
- One IFO measures a single linear combination of h<sub>+</sub> and h×. A network of misaligned IFOs is needed to separate the polarizations and to localize (arrival-time delays up to ≈ 10 ms for Hanford–Livingston).
- In a *targeted* analysis the sky position is supplied externally, so antenna factors and time delays are fixed. This, together with the restricted time window, is why targeted searches are more sensitive than all-sky ones.
- The paper summarizes the purely geometric response of the network by $F_{\mathrm{net}} = [\Sigma _{i} (F_{+},_{i}^{2} + F_{\times},_{i}^{2})]^{(1/2)}$ (Sect. 3.3); it carries no noise weighting.

### 1.3 Generation: the quadrupole formula

At leading order GWs are sourced by the time-varying mass quadrupole,

$$
\begin{gathered}
h_{\mathrm{ij}}^{\mathrm{TT}}(t, r) = (2G / c^{4}r) \cdot  d^{2}Q_{\mathrm{ij}}^{\mathrm{TT}}/dt^{2} (t - r/c),\quad Q_{\mathrm{ij}} = \int  \rho  x_{i} x_{j} d^{3}x
\end{gathered}
$$

The coupling G/c<sup>4</sup> ≈ 8×10<sup>-45</sup> s<sup>2</sup> kg<sup>-1</sup> m<sup>-1</sup> makes spacetime extraordinarily stiff: only compact, massive, relativistic and non-axisymmetric systems produce measurable strain, h $\sim$ (G/c<sup>4</sup>) $E_{\mathrm{kin}}$^ns / r up to factors of order unity, with $E_{\mathrm{kin}}$^ns the non-spherical kinetic energy.

### 1.4 Quasi-circular binaries: chirp mass and inspiral

For a quasi-circular binary the GW frequency is twice the orbital frequency. At leading order (Newtonian orbit, quadrupole radiation reaction):

$$
\begin{gathered}
\mathcal{M}  = (m_{1}m_{2})^{(3/5)} / (m_{1}+m_{2})^{(1/5)}\quad\text{chirp mass} \\
\dot{f} = (96/5) \pi ^{(8/3)} (G\mathcal{M} /c^{3})^{(5/3)} f^{(11/3)}\quad\text{frequency evolution} \\
\tau (f) = (5/256) (G\mathcal{M} /c^{3})^{(-5/3)} (\pi f)^{(-8/3)}\quad\text{time to coalescence}
\end{gathered}
$$

For an observer at inclination ι (angle between the line of sight and the orbital angular momentum) and luminosity distance $D_{\mathrm{L}}$,

$$
\begin{gathered}
h_{+} = (4/D_{\mathrm{L}}) (G\mathcal{M} /c^{2})^{(5/3)} (\pi f/c)^{(2/3)} \cdot  \frac{1}{2}(1+\cos^{2}\iota ) \cdot  \cos\Phi (t) \\
h_{\times} = (4/D_{\mathrm{L}}) (G\mathcal{M} /c^{2})^{(5/3)} (\pi f/c)^{(2/3)} \cdot  \cos\iota  \cdot  \sin\Phi (t)\quad \Phi (t) = 2\pi  \int  f dt
\end{gathered}
$$

Consequences:

1. At leading order the masses enter only through ℳ: amplitude ∝ $\mathcal{M} ^{(5/3)}$ $f^{(2/3)}$ and ḟ ∝ $\mathcal{M} ^{(5/3)}$.
2. The polarization is circular for ι = 0 (face-on) and linear for ι = 90° (edge-on). The sky- and polarization-averaged SNR scales as √\[(1+cos<sup>2</sup>ι)<sup>2</sup> + 4cos<sup>2</sup>ι\] (Finn & Chernoff 1993, quoted in Sect. 2.1 of the paper), i.e. a factor √8 ≈ 2.8 between face-on and edge-on.
3. GRB jets are collimated (typical half-opening angles $\sim$ 10°, larger values not excluded) and the jet axis is expected to be aligned with the orbital angular momentum, so an observed GRB implies a small ι. This is the origin of the paper's conservative EM prior ι ≤ 45° (Sect. 2.1).
4. The signal is long-lived, with 10<sup>2</sup>–10<sup>3</sup> cycles in band, so the SNR is accumulated coherently over many cycles. This is what matched filtering exploits.

**Worked numbers (derived).** For 1.4 + 1.4 M⊙: ℳ = 1.22 M⊙ and Gℳ/c<sup>2</sup> ≈ 1.8 km. Starting at 30 Hz (lower edge of the band used in the paper, Sect. 2.2) the signal lasts τ ≈ 53 s and sweeps ≈ 2.6×10<sup>3</sup> GW cycles up to 1024 Hz; a face-on source at 100 Mpc has h ≈ 3.6×10<sup>-23</sup> at f = 100 Hz.

> **Takeaway.** The detector response reduces to two numbers per IFO, `F₊` and $F_{\times}$, which depend only on the sky position and ψ; the EM trigger supplies the sky position. The source enters through ℳ, the distance and the inclination.

## 2. Compact-binary waveforms (9 min)

### 2.1 Frequency-domain inspiral: stationary phase and TaylorF2

Matched filtering is carried out in the Fourier domain. For a slowly chirping signal the stationary-phase approximation (SPA) gives |h̃(f)| = A($t_{\mathrm{f}}$) / (2√ḟ), where $t_{\mathrm{f}}$ is the instant at which the frequency equals f. Inserting the Newtonian amplitude and ḟ from §1.4, for an optimally oriented source (`F₊` = 1, $F_{\times}$ = 0, ι = 0):

$$
\begin{gathered}
|\tilde{h}(f)|^{2} = (5/24) \pi ^{(-4/3)} (G\mathcal{M} )^{(5/3)} c^{-3} D_{\mathrm{L}}^{-2} f^{(-7/3)}
\end{gathered}
$$

For general geometry $D_{\mathrm{L}}$ is replaced by the effective distance $D_{\mathrm{eff}}$ (§4). The Fourier amplitude falls as $f^{(-7/6)}$. The phase, to 1.5PN order (TaylorF2 family), is

$$
\begin{gathered}
\Psi (f) = 2\pi f t_{c} - \phi _{c} - \pi /4 + (3/128\eta ) v^{-5} [1 + (3715/756 + 55\eta /9) v^{2} - 16\pi  v^{3} + \ldots ] \\
v = (\pi GMf/c^{3})^{(1/3)}
\end{gathered}
$$

- The leading term depends on ℳ only: (3/128)(πGℳf/c<sup>3</sup>)^(−5/3).
- 1PN: mass ratio; 1.5PN: tail term (−16π) and spin–orbit; 2PN: spin–spin.
- Tides enter at 5PN: $\delta \Psi _{\mathrm{tidal}} = -(117/256\eta ) \tilde{\Lambda } v^{5}$, with $\Lambda _{i} = (2/3) k_{2},i (R_{i} c^{2}/G m_{i})^{5}$ and $\tilde{\Lambda } = (16/13) [(m_{1}+12m_{2}) m_{1}^{4} \Lambda _{1} + (m_{2}+12m_{1}) m_{2}^{4} \Lambda _{2}] / M^{5}$. Tidal deformation accelerates the inspiral, matters above a few hundred Hz, and encodes the equation of state (EOS): a stiff EOS (large R) gives a large Λ. For a black hole Λ = 0. The paper (Sect. 2.1.2, Table 1) obtains Λ of the NS from its compactness through the universal relation of Godzieba et al. (2021).
- The inspiral expansion ends near the innermost stable circular orbit, $f_{\mathrm{ISCO}} = c^{3} / (6^{(3/2)} \pi  G M) \approx  4.4 kHz \times  (M_{\odot}/M)$.

*Why phase accuracy matters.* A phase error of order 1 rad accumulated across the band destroys a large fraction of the SNR; with ≈ 10<sup>3</sup> cycles the matched filter is an extremely sensitive phase detector. Residual waveform systematics are therefore absorbed in a mismatch factor k (§3.4).

### 2.2 Merger and ringdown; the six mass pairs of the paper

For heavy systems the merger lies inside the detector band and inspiral-only templates are insufficient. The paper uses TaylorF2 (with tides) for BNS and IMRPhenomNSBH for NSBH (Sect. 2.2), a non-precessing inspiral–merger–ringdown model that includes the NS tidal disruption. NS spin and BH tidal deformability are set to zero. The six default mass pairs and their properties are:

| Pair \[m<sub>1</sub>, m<sub>2</sub>\] (M⊙) | ℳ (M⊙, derived) | $f_{\mathrm{ISCO}}$ (Hz, derived) | Minimum $\chi _{\mathrm{BH}}$ for $M_{\mathrm{REM}}$ > 0 (paper) | Waveform (paper) |
| --- | --- | --- | --- | --- |
| BNS \[1, 1\] | 0.87 | ≈ 2200 | — | TaylorF2 + tides |
| BNS \[1.4, 1.4\] | 1.22 | ≈ 1570 | — | TaylorF2 + tides |
| BNS \[2, 2\] | 1.74 | ≈ 1100 | — | TaylorF2 + tides |
| NSBH \[1, 5\] | 1.84 | ≈ 730 | any spin (including 0) | IMRPhenomNSBH |

| NSBH \[2, 20\] | 4.93 | ≈ 200 | ≈ 0.94–0.99 | IMRPhenomNSBH |

For the BNS pairs $f_{\mathrm{ISCO}}$ lies at or above the upper edge of the analysis band ($f_{\mathrm{H}}$ = 1024 Hz); for all NSBH pairs the merger falls inside the band. The six pairs span ℳ ≈ 0.9–4.9 M⊙, i.e. the widest range of detectability for the EM-bright scenarios considered.

### 2.3 What makes a merger EM-bright? (paper, Sects. 2.1.1–2.1.2)

The working hypothesis is that the CBC powers the EM transient, which requires a non-zero *remnant mass* $M_{\mathrm{REM}}$ of baryons left outside the compact object formed in the merger (not to be confused with the mass of that object). It may be bound (a hyper-accreting disk) or unbound: dynamical ejecta, neutrino-driven winds, viscous disk outflows, magnetically driven outflows and, for a fast-rotating magnetar remnant, a particle wind. The same ejecta open the way to neutrino and cosmic-ray production.

- **NSBH.** $M_{\mathrm{REM}}$ depends on the mass ratio, the NS EOS and the BH spin (the paper uses the fitting formulae of Foucart et al. 2018, and the soft SFHo and stiff DD2 EOSs). Order-of-magnitude picture: the NS is tidally disrupted at $r_{t} \approx  R_{\mathrm{NS}} (m_{\mathrm{BH}}/m_{\mathrm{NS}})^{(1/3)}$ and leaves matter outside the horizon only if $r_{t}$ exceeds $r_{\mathrm{ISCO}}$. For a fixed NS, $r_{t}$ grows slowly with $m_{\mathrm{BH}}$, while $r_{\mathrm{ISCO}} \propto  m_{\mathrm{BH}}$ and $r_{\mathrm{ISCO}} / (G m_{\mathrm{BH}}/c^{2})$ falls from 6 (χ = 0) to 1 (χ → 1, prograde orbit). Heavier BHs therefore need larger spins, which is the monotonic relation of the paper's Fig. 1 and the logic behind the three NSBH pairs. The \[2, 20\] pair, requiring $\chi _{\mathrm{BH}}$ ≈ 0.94–0.99, is considered more exotic by the authors but still viable.
- **BNS.** EM-brightness depends on the total mass (prompt collapse versus a massive or hypermassive NS), the mass ratio (more asymmetric systems give heavier disks and dynamical ejecta), and the remnant (a magnetar can brighten the kilonova and power winds or plateaus). Given this plethora of scenarios the paper applies no stringent mass cut and brackets the chirp-mass range with equal-mass binaries.

> **Takeaway.** For detectability the waveform matters mainly through ℳ (amplitude ∝ $\mathcal{M} ^{(5/3)}$ at fixed f), the high-frequency termination ($f_{\mathrm{ISCO}}$ or tidal disruption) and the tidal phase. The EM-bright requirement selects the astrophysically relevant mass pairs and, for NSBH, a minimum BH spin.

## 3. Noise, matched filtering and signal-to-noise ratio (11 min)

### 3.1 Data model and PSD

$$
\begin{gathered}
s(t) = n(t) + h(t) \\
\langle \tilde{n}(f) \tilde{n}^{*}(f^{\prime})\rangle  = \frac{1}{2} S_{n}(f) \delta (f - f^{\prime})
\end{gathered}
$$

Noise is modelled as zero-mean, stationary and Gaussian, fully characterized by the one-sided PSD $S_{n}(f)$. For the O1–O3 detectors $S_{n}$ is dominated by seismic noise below ≈ 20 Hz, thermal noise at tens to a few hundred Hz, and quantum shot noise above ≈ 200 Hz; a BNS signal gathers most of its SNR between tens and a few hundred Hz. Real data are neither exactly stationary nor Gaussian (spectral lines, glitches, slow drifts).

*Estimating $S_{n}$.* Welch’s method splits the time series into overlapping windowed segments, Fourier transforms each, and averages the periodograms, trading frequency resolution for variance. The paper uses 16 s segments with 50% overlap (Sect. 2.2). An IFO is counted as observing only if it has continuous data over `[t₀ − 128 s, t₀ + 128 s]`, which the authors take as the minimum span for a reliable noise estimate. Practical constraint: LVK releases only day-averaged PSDs in real time, so the TDR is currently run offline on GWOSC strain; a genuinely low-latency version requires a PSD stream (the paper notes that the online implementation itself is immediate).

*Check yourself.* How many overlapping 16 s segments fit in a 256 s span? (31, derived.)

### 3.2 Inner product, optimal SNR, matched filter

$$
\begin{gathered}
(a|b) = 4 \Re  \int _{f_{\mathrm{L}}}^{f_{\mathrm{H}}} \tilde{a}(f) \tilde{b}^{*}(f) / S_{n}(f) df\quad\text{(paper Eq. 9)} \\
\rho _{\mathrm{opt}} = (h|h)^{(1/2)} = [ 4 \int _{f_{\mathrm{L}}}^{f_{\mathrm{H}}} |\tilde{h}(f)|^{2} / S_{n}(f) df ]^{(1/2)}\quad\text{(Eqs. 1, 7)}
\end{gathered}
$$

The paper takes $f_{\mathrm{L}}$ = 30 Hz and $f_{\mathrm{H}}$ = 1024 Hz, the values of the LVK GRB-targeted searches; the contribution outside this range is negligible for current detectors, and the interval must be widened for future ones. The inner product weights frequencies by the inverse noise: $\rho _{\mathrm{opt}}^{2}$ is the signal power per unit noise power integrated over the band.

If the waveform h is known exactly, the statistic $\rho  = (s|\hat{h})$ with $\hat{h} = h/(h|h)^{(1/2)}$ is Gaussian with unit variance and mean 0 (noise only) or $\rho _{\mathrm{opt}}$ (signal present). It maximizes the SNR among linear filters (Wiener filter) and is the Neyman–Pearson statistic, $ln \Lambda  = (s|h) - \frac{1}{2}(h|h)$. Hence $\rho _{\mathrm{opt}}$ is the *expected* matched-filter SNR for a perfectly modelled signal: it depends only on the signal model and the PSD, not on the noise realization.

### 3.3 Unknown phase: the matched-filter SNR and its statistics

The coalescence phase is unknown. With two templates in quadrature, $h_{\mathrm{I}}$ and $h_{\mathrm{Q}}$ (90° out of phase), the phase-maximized statistic is (paper Eq. 8)

$$
\begin{gathered}
\rho _{\mathrm{MF}} = [ (s|h_{\mathrm{I}})^{2} + (s|h_{\mathrm{Q}})^{2} ]^{(1/2)} / (h_{\mathrm{I}}|h_{\mathrm{I}})^{(1/2)}
\end{gathered}
$$

Over noise realizations $\rho _{\mathrm{MF}}^{2}$ follows a χ<sup>2</sup> law with 2 degrees of freedom: central (mean 2) without a signal, non-central with non-centrality parameter $k \rho _{\mathrm{opt}}^{2}$ with a signal, where k ≤ 1 accounts for template–signal mismatch (§3.4). For a network of $N_{\mathrm{IFO}}$ IFOs with independent noise, $\rho _{\mathrm{net}}^{2} = \Sigma _{i} \rho _{i}^{2}$ (Eq. 5) and the distribution has 2$N_{\mathrm{IFO}}$ degrees of freedom (Sect. 2.3). Useful moments (standard non-central χ<sup>2</sup> results):

$$
\begin{gathered}
\langle \rho _{\mathrm{MF}}^{2}\rangle  = 2 N_{\mathrm{IFO}} + k \rho _{\mathrm{net}}^{2},\quad Var(\rho _{\mathrm{MF}}^{2}) = 4 N_{\mathrm{IFO}} + 4 k \rho _{\mathrm{net}}^{2} \\
\rho _{\mathrm{opt}} \gg  1 :\quad \rho _{\mathrm{MF}} \approx  \sqrt{k} \rho _{\mathrm{opt}} + \mathcal{N} (0, 1)
\end{gathered}
$$

*Example (derived).* $N_{\mathrm{IFO}}$ = 2, $\rho _{\mathrm{net}}$ = 10, k = 0.95 give ⟨$\rho _{\mathrm{MF}}$<sup>2</sup>⟩ = 4 + 95 = 99, i.e. a typical $\rho _{\mathrm{MF}}$ ≈ 9.9 with ≈ 1 unit of scatter.

### 3.4 Templates, mismatch and the factor k

The *match* between two normalized waveforms is their overlap maximized over time and phase. A template bank is laid out so that every signal in the target space has match ≥ MM (typically 0.97) with some template. Because the recovered SNR scales with the match, the loss in SNR<sup>2</sup> is up to 1 − MM<sup>2</sup> ≈ 6% *(derived)*, comparable with the paper’s choice `k ~ U[0.9, 1]` (Sect. 2.3), which also covers waveform systematics. The authors note (Sect. 3.2) that the actual template–signal mismatch is more complex than this single factor.

### 3.5 From SNR to significance

In Gaussian noise the single-trial false-alarm probability for a single-IFO threshold is $exp(-\rho _{c}^{2}/2)$. Real noise has non-Gaussian tails, so pipelines apply signal-consistency tests, build a ranking statistic, and measure the background empirically (time slides in all-sky searches; off-source data in targeted ones). The outputs are the false-alarm rate (FAR) and the probability of astrophysical origin $p_{\mathrm{astro}}$. Fig. 2 of the paper uses all candidates in GWTC-2.1, GWTC-3 and GWTC-4: among those with matched-filter SNR > 10, ≈ 60% have FAR < 1 yr<sup>-1</sup> and $p_{\mathrm{astro}}$ > 0.5. This coarse SNR-to-significance mapping motivates the $\rho _{\mathrm{cut}}$ = 10 used for the paper’s Table 3. The mapping depends on pipeline, mass region and epoch; hence $\rho _{\mathrm{cut}}$ is a user choice.

> **Takeaway.** The *optimal* SNR is deterministic given signal and PSD; the *matched-filter* SNR is a random variable, scattered by noise and degraded by mismatch. The TDR works with the former and converts to the latter through a non-central χ<sup>2</sup> law.

## 4. Effective distance, horizon and range (7 min)

### 4.1 Geometry only rescales the amplitude

For a single dominant harmonic the antenna-weighted signal is

$$
\begin{gathered}
h(t) = A(t) [ F_{+} \frac{1}{2}(1+\cos^{2}\iota ) \cos\Phi  + F_{\times} \cos\iota  \sin\Phi  ] = A(t) \Omega  \cos(\Phi  - \Phi _{0}) \\
\Omega  = [ F_{+}^{2} ((1+\cos^{2}\iota )/2)^{2} + F_{\times}^{2} \cos^{2}\iota  ]^{(1/2)}  \leq  1
\end{gathered}
$$

with `A(t)` the face-on amplitude of §1.4 and Φ<sub>0</sub> a constant phase. Sky position, polarization angle and inclination therefore only rescale the amplitude and shift a constant phase. Defining the effective distance (paper Eq. 3),

$$
\begin{gathered}
D_{\mathrm{eff}} = D_{\mathrm{L}} / \Omega  = D_{\mathrm{L}} [ F_{+}^{2} ((1+\cos^{2}\iota )/2)^{2} + F_{\times}^{2} \cos^{2}\iota  ]^{(-1/2)}  \geq   D_{\mathrm{L}}
\end{gathered}
$$

the optimal SNR scales as $\rho _{\mathrm{opt}} \propto  1/D_{\mathrm{eff}}$. This is the identity behind the paper’s rescaling (Eqs. 2 and 4):

$$
\begin{gathered}
\rho (RA, Dec, D_{\mathrm{L}}, \iota , \psi ) = \rho _{\mathrm{ref}} \cdot  D_{\mathrm{eff,ref}} / D_{\mathrm{eff}}(RA, Dec, D_{\mathrm{L}}, \iota , \psi ) \\
D_{\mathrm{eff,ref}} = D_{\mathrm{eff}}(RA_{\mathrm{ref}}, Dec_{\mathrm{ref}}, D_{L,\mathrm{ref}}, \iota  = 0, \psi  = 0)
\end{gathered}
$$

One waveform computation per IFO, per mass pair and per EOS yields $\rho _{\mathrm{ref}}$; every other geometry follows analytically. This is the origin of the TDR’s low computational cost. The identity holds for waveforms dominated by a single (ℓ = 2, |m| = 2) harmonic and without precession, as is the case for the models adopted.

### 4.2 Optimal SNR in terms of ℳ and the PSD

Combining the SPA amplitude of §2.1 with the definition of $\rho _{\mathrm{opt}}$ (inspiral only, leading-order amplitude):

$$
\begin{gathered}
\rho _{\mathrm{opt}}^{2} = (5/6) \pi ^{(-4/3)} (G\mathcal{M} )^{(5/3)} c^{-3} D_{\mathrm{eff}}^{-2} \int _{f_{\mathrm{L}}}^{f_{\mathrm{H}}} f^{(-7/3)} / S_{n}(f) df
\end{gathered}
$$

Reading: (i) $\rho  \propto  \mathcal{M} ^{(5/6)}$; (ii) $\rho  \propto  1/D_{\mathrm{eff}}$; (iii) the integrand $f^{(-7/3)}/S_{n}(f)$ peaks where the detector is most sensitive, at tens to a few hundred Hz; (iv) for heavy systems the integral is truncated at $f_{\mathrm{ISCO}}$ (and merger–ringdown corrections apply), which reduces the gain from a larger ℳ. Naive inspiral-only scaling *(derived)*: across the six mass pairs ℳ spans 0.87–4.93 M⊙, a factor ≈ 4.2 in reach and ≈ 76 in volume before truncation effects.

### 4.3 Horizon, range and the factor 2.26

The horizon distance is the largest distance at which an optimally located and oriented binary ($D_{\mathrm{eff}} = D_{\mathrm{L}}$) reaches the threshold,

$$
\begin{gathered}
D_{h} = \rho _{\mathrm{opt}}(D_{\mathrm{eff}} = 1) / \rho _{\mathrm{thr}}
\end{gathered}
$$

A binary at $D_{\mathrm{L}}$ is detectable ($\rho _{\mathrm{opt}} > \rho _{\mathrm{thr}}$) if and only if $D_{\mathrm{L}} < D_{h} \Omega $. The sensitive volume is $V = (4\pi /3) D_{h}^{3} \langle \Omega ^{3}\rangle $ (average over sky, polarization and cosι), and the *range* R is defined by $V = (4\pi /3) R^{3}$, i.e. $R = D_{h} \langle \Omega ^{3}\rangle ^{(1/3)}$. For a single IFO this gives $D_{h} / R \approx  2.26$, i.e. $\langle \Omega ^{3}\rangle ^{(1/3)} \approx  0.44$ (Finn & Chernoff 1993; Allen et al. 2012), the factor quoted in Sect. 1 of the paper. The real-time LVK *BNS range* uses a 1.4 + 1.4 M⊙ system and, conventionally, $\rho _{\mathrm{thr}}$ = 8.

*Worked example (derived).* For R = 130 Mpc (the order of the O3 LIGO ranges), $D_{h}$ ≈ 294 Mpc, so an overhead face-on BNS at 100 Mpc has $\rho _{\mathrm{opt}}$ ≈ 8 × 294/100 ≈ 24 in that IFO.

### 4.4 Range versus TDR

The range is a property of the detector; the TDR is a property of the transient.

| Aspect | BNS range | TDR |
| --- | --- | --- |
| Masses | 1.4 + 1.4 M⊙ | Three BNS and three NSBH pairs (EM-bright) |
| Sky position | Averaged over the sphere | Localization map of the trigger (or fixed RA/Dec) |
| Inclination | Isotropic | EM prior: 0–45° (jet) or 0–90° (isotropic) |
| Noise | Current PSD of each IFO | PSD of each IFO observing at the trigger time |
| Threshold | $\rho _{\mathrm{thr}}$ = 8 (convention) | $\rho _{\mathrm{cut}}$ ≈ 9–10, user-adjustable |
| Criterion | Volume average | 90% detection efficiency (λ adjustable) |
| Output | One number per IFO | Efficiency curve, D<sub>90</sub>, D<sub>90</sub> versus sky position |

> **Takeaway.** ρ ∝ $\mathcal{M} ^{(5/6)}$/$D_{\mathrm{eff}}$: once $\rho _{\mathrm{ref}}$ is known, any sky position, inclination and distance costs one line of algebra. The TDR is the range re-evaluated for the specific transient.

## 5. Searches, significance and exclusion distances (6 min)

### 5.1 All-sky versus targeted

An all-sky, all-time search scans the whole sky, all times and a broad parameter space; the large number of trials raises the detection threshold. An external trigger fixes the time window, the sky position and part of the parameter space (inclination, masses), allowing a lower threshold at equal false-alarm rate. Within LVK the targeted CBC search is PyGRB (coherent matched filtering; Harry & Fairhurst 2011; Williamson et al. 2014) and the unmodelled one is X-Pipeline (Sutton et al. 2010; Was et al. 2012). Latency is of the order of hours; the targeted GstLAL version of Tsukada et al. (2026) is ≈ 50 times faster than the all-sky one (paper, Sect. 1).

### 5.2 Exclusion distances

In a targeted search the background is estimated from off-source data, and the loudest on-source event is compared with it. If it is not significant, upper limits follow from an injection campaign: simulated signals are added to real data and recovered with the same pipeline. The **90% exclusion distance** is the distance at which 90% of the injections are recovered with a ranking statistic larger than the maximum observed in the on-source window (paper, Sects. 2.2 and 3.1; LVK GRB papers for O1–O3). Reading: had a merger from the assumed population occurred closer than `D₉₀`, there would have been at least a 90% chance of observing a louder event; its absence therefore disfavors, in the frequentist sense, a CBC origin closer than `D₉₀` for the assumed priors.

- The value depends on the injection population (masses, inclination range) and on the noise at the time (antenna response, which IFOs observe).
- Bayesian alternatives (Sinha et al. 2026: parameter estimation and evidence, uniform-in-volume distance prior) give bounds that depend on the distance prior. In the paper’s comparison (Sect. 3.3) they are systematically larger than the TDR because the prior weights large distances more heavily, so they are not directly comparable with frequentist exclusion distances.

### 5.3 Why a TDR can stand in for an exclusion distance

If no pipeline finds a significant candidate, the loudest on-source statistic sits at the level of the noise background, and a signal with SNR above $\rho _{\mathrm{cut}}$ ≈ 9–10 would have been significant. Replacing the on-source loudest event by a fixed SNR threshold, and the injection campaign by simulated SNRs, yields the TDR. The substitution neglects glitches, details of the ranking statistic and on-source data quality, which is why the paper calibrates $\rho _{\mathrm{cut}}$ empirically against PyGRB (§6.4).

> **Takeaway.** An exclusion distance is a statement about a null result and an assumed signal population. The TDR approximates it without touching the data, apart from the PSD.

## 6. The targeted detectability range (12 min)

### 6.1 Definition (paper Sect. 2.2, Eq. 6)

For a chosen threshold $\rho _{\mathrm{cut}}$ and efficiency λ, the EM-informed detectability range $D_{\lambda |\rho _{\mathrm{cut}}}$ is the distance at which

$$
\begin{gathered}
P(\rho _{\mathrm{net}} > \rho _{\mathrm{cut}}) = \lambda 
\end{gathered}
$$

that is, a fraction λ of simulated sources drawn from the EM-informed priors is recovered above threshold. Because ρ decreases monotonically with distance at fixed geometry, the efficiency $\epsilon (D) = P(\rho _{\mathrm{net}} > \rho _{\mathrm{cut}} | D)$ decreases monotonically and the solution is unique. The **targeted detectability range** is $D_{90}^{\mathrm{TDR}}$ (λ = 0.9, chosen for consistency with the LVK exclusion distances). Operationally (Sect. 2.3): given an external messenger assumed to be driven by a CBC, there is a 90% probability that a GW search pipeline would recover the signal at $D_{90}^{\mathrm{TDR}}$ with a measured SNR ≥ $\rho _{\mathrm{cut}}$. The tool also returns the full efficiency curve ε(D), so that any other λ can be adopted.

### 6.2 Inputs supplied by the EM observation (Sects. 2.1, 2.1.3)

1. **Trigger time** `t₀` and the IFOs observing at `t₀`.
2. **Sky localization**: from arcseconds–arcminutes (optical, X-ray; RA/Dec kept fixed) to tens–hundreds of square degrees (Fermi-GBM, IPN triangulation); in the latter case the injected positions are drawn from the probability map.
3. **Inclination prior**: isotropic (uniform in cosι) over $[0, \pi /4]$ or $[0, \pi /2]$, chosen according to the trigger:

| Trigger | Prior | Rationale (paper) |
| --- | --- | --- |
| GRB (relativistic jet) | ι ∈ \[0°, 45°\] | Jet-break angles $\sim$ 10°; larger values allowed but unmeasured owing to detection biases |
| Kilonova-like transient | Isotropic | Outflows expected to be nearly isotropic; no strong prior |
| Soft, fast X-ray transient | Narrower if on- or mildly off-axis jet; isotropic if cocoon or magnetar wind | Depends on interpretation; MeV counterpart and multiwavelength data help discriminate |
| Neutrino, jet-like origin | ι ∈ \[0°, 45°\] | GRB case |
| Neutrino, choked jet (no GRB) | ι ∈ \[0°, 90°\] | Isotropic prior |
| No EM counterpart | Run all mass pairs and both priors | Most agnostic option |

4. **Masses**: the six EM-bright pairs of §2.2, with $\chi _{\mathrm{BH}}$ set to the minimum value giving $M_{\mathrm{REM}}$ > 0 and the NS Λ from Table 1.

For neutrino triggers the sub-degree localization of tracks above a few tens of TeV makes the sky-position uncertainty negligible; the main complication is the uncertain delay (seconds to days) between merger and neutrino emission, handled by evaluating the TDR on several time segments.

### 6.3 The algorithm (Sect. 2.2)

1. **IFO status.** An IFO is included if it has data over `[t₀ − 128 s, t₀ + 128 s]` (`dq.query_flag` from PyCBC).
2. **PSD.** Welch estimate from the strain data, 16 s segments, 50% overlap.
3. **Reference SNR.** For each mass pair (and EOS, for NSBH) generate the waveform (TaylorF2 or IMRPhenomNSBH, via `pycbc_create_injections`) and compute $\rho _{\mathrm{ref}}$ per IFO with `pycbc_optimal_snr` over 30–1024 Hz.
4. **Monte Carlo geometry.** Draw the sky position from the localization map (or fix it), ι from the prior (uniform in cosι), ψ and $\phi _{\mathrm{c}}$ uniformly; compute $D_{\mathrm{eff}}$ for each IFO (Eq. 3).
5. **Distance scan.** For each trial distance, $\rho _{i} = \rho _{\mathrm{ref}} D_{\mathrm{eff,ref}} / D_{\mathrm{eff}}$ (Eq. 2) and $\rho _{\mathrm{net}} = (\Sigma _{i} \rho _{i}^{2})^{(1/2)}$ (Eq. 5).
6. **Optimal to matched-filter SNR.** Draw k $\sim$ U\[0.9, 1\] and a non-central χ<sup>2</sup> variate with 2$N_{\mathrm{IFO}}$ degrees of freedom and non-centrality $k \rho _{\mathrm{net}}^{2}$; the efficiency is the fraction with $\rho _{\mathrm{MF}} > \rho _{\mathrm{cut}}$.
7. **Output.** ε(D) for each pair and prior, $D_{90}^{\mathrm{TDR}}$, and `D₉₀` as a function of sky position (paper Fig. 6).

The paper quotes a run time of minutes (Sect. 5). Remarks:

- The sky average is a probability-weighted average over the localization map: `D₉₀` is a property of the map, not of its peak. The map peak is used in the paper only to attach a single $F_{\mathrm{net}}$ value to each GRB in Fig. 5.
- Only step 3 requires waveform generation; everything else is cheap algebra and Monte Carlo sampling.
- If the merger time is uncertain, steps 1–7 are repeated on successive time segments around `t₀` (Sect. 5, conclusions).

### 6.4 Validation against PyGRB (Sects. 3.1–3.2)

**Sample.** 574 GRBs detected by Fermi-GBM and Swift-BAT fall in the O1–O3 time windows (Swift preferred when both detect, for its better localization); 467 have at least one IFO observing. No cut on GRB duration is applied.

**Comparison set.** PyGRB searched GRBs classified as short or ambiguous. Of the 82 Fermi and Swift GRBs analysed in the LVK O1, O2, O3a and O3b papers, 71 have the same IFO set as in the TDR calculation (differences arise when LVK added an IFO outside nominal observing mode or removed one for data-quality reasons). Only BNS are compared: for NSBH the BH mass distributions and EM-bright prescriptions differ, with systematics that are hard to control.

**Matched settings.** ι ≤ 30° (instead of the default 45°, as in the LVK injections) and 1.4 + 1.4 M⊙ (PyGRB draws NS masses from a Gaussian with mean 1.4 and standard deviation 0.2 M⊙; the authors find a negligible bias).

**Metric and calibration** (Eq. 10):

$$
\begin{gathered}
\Delta _{90}(\rho _{\mathrm{cut}}) = [ D_{90}^{\mathrm{TDR}}(\rho _{\mathrm{cut}}) - D_{90}^{\mathrm{PyGRB}} ] / D_{90}^{\mathrm{TDR}}(\rho _{\mathrm{cut}})
\end{gathered}
$$

The mean of |Δ<sub>90</sub>| over GRBs is minimized with respect to $\rho _{\mathrm{cut}}$ (step 0.5), with the optimum at $\rho _{\mathrm{MF}}$ = 9, equivalently $\rho _{\mathrm{opt}}$ = 10.

**Result.** At $\rho _{\mathrm{MF,cut}}$ = 9 the median of Δ<sub>90</sub> is 0.02 and the 16–84% range is \[−0.14, +0.20\]; a non-negligible subset shows 20–60% differences. The conclusions summarize this as a relative mismatch ≲ 20% for ≈ 70% of the GRBs. The authors attribute the residuals to (i) non-stationary, non-Gaussian noise, (ii) differences between injection populations, and (iii) a template–signal mismatch richer than the single factor k.

The optimum for $\rho _{\mathrm{opt}}$ exceeding that for $\rho _{\mathrm{MF}}$ by about one unit is in the expected direction, since noise scatter (≈ 1, §3.3) and mismatch (§3.4) raise the optimal SNR needed for 90% efficiency (interpretation, not stated in the paper).

**Cross-checks (Sect. 3.3).** Wang et al. (2022; long GRBs in O3, PyCBC; exclusion distance defined at optimal SNR = 8 and ι = 30°): agreement within a few percent when the same injections are assumed. Sinha et al. (2026; Bayesian evidence, long GRBs): their 90% exclusion distances are systematically larger, which the authors ascribe to the uniform-in-volume distance prior.

### 6.5 Results for O1–O3 (Sects. 3.3–3.4)

- For 1.4 + 1.4 M⊙, ι ∈ \[0°, 30°\] and $\rho _{\mathrm{MF,cut}}$ = 9, the median $D_{90}^{\mathrm{TDR}}$ rises from ≈ 70 Mpc in O1 to ≈ 140 Mpc at the end of O3, tracking the evolution of the sky-averaged BNS range.
- It correlates strongly with the network antenna factor $F_{\mathrm{net}}$ (position relative to the antenna patterns) and with the number of IFOs observing.
- The highest value in the sample is GRB 200228291, whose localization sits around a peak of the network antenna pattern.
- Table 3 lists $D_{90}^{\mathrm{TDR}}$ for BNS and NSBH at $\rho _{\mathrm{cut}}$ = 10 (the more conservative choice), with the PyGRB values where available; the full table is on Zenodo.
- Standard products (Fig. 6): efficiency versus distance for each mass pair and inclination prior, and `D₉₀` versus sky position with the localization contours and injected positions.

### 6.6 Science cases (Sect. 4)

1. **Prioritizing follow-up.** Compare the candidate-host distance with the GW exclusion distance: a host inside it, with no GW candidate, makes a merger origin unlikely; a more distant host leaves it viable and justifies continued effort if the goal is to find merger-driven events. *GRB 250225B* (Swift-BAT; galaxy at 78 Mpc inside the arcminute localization, chance-coincidence probability < 1%): a prompt estimate would have strengthened or weakened the merger interpretation, and a non-detection despite a high detection probability would have strongly disfavored the host. *GRB 251026B* (Fermi-GBM; GOTO candidate counterpart in a galaxy at ≈ 250 Mpc, later classified as a Type Ia supernova): the TDR would have rapidly disfavored a merger.
2. **Interpreting the origin of transients.** *GRB 211211A* (host at ≈ 350 Mpc) and *GRB 230307A* (≈ 290 Mpc) are long-duration GRBs with kilonova-like features that lie within the LVK horizon, especially for NSBH progenitors; the TDR gives a rapid GW-detectability assessment to single out merger-driven events among long GRBs. The same applies to fast X-ray transients discovered by Einstein Probe, treated systematically for the second half of O4 in the companion paper (Chopra et al. 2026).
3. **Unknown merger time.** *AT2025ulz* (ZTF transient coincident with the low-significance candidate S250818k; later classified as a Type II supernova, also proposed as a superkilonova) and *orphan afterglows* (jet-launch time from light-curve modelling, uncertainty as small as ≈ 200–300 s for well-sampled cases): evaluate the TDR on all time segments within the uncertainty window, to disfavor or leave open the merger channel.

### 6.7 Limitations acknowledged by the authors

- The TDR is not a search: $D_{90}^{\mathrm{TDR}}$ is not rigorously an exclusion distance and approximates one only when no pipeline finds a significant candidate in the window. The offline exclusion distance of targeted searches such as PyGRB remains “the most accurate and reliable metric” (Sect. 3.2).
- Non-stationary, non-Gaussian noise and waveform mismatch enter only through the factor k.
- Component masses are fixed; the NSBH case is not validated against PyGRB.
- Low-latency operation requires real-time PSDs (currently only day-averaged ones are released).
- $\rho _{\mathrm{cut}}$ is representative and adjustable; the band $f_{\mathrm{L}}$–$f_{\mathrm{H}}$ must be extended for third-generation detectors.

### 6.8 Points for critical discussion (remarks added for these notes, not claims of the paper)

1. **Calibration versus validation.** $\rho _{\mathrm{cut}}$ = 9 is the value minimizing ⟨|Δ<sub>90</sub>|⟩ on the very sample used for the comparison, so the small median offset (0.02) is partly by construction. The informative quantities are the scatter (16–84%: −0.14 to +0.20) and the tail (20–60%). An out-of-sample test (cross-validation across observing runs, or comparison with O4 targeted-search results) would strengthen the claim.
2. **The \[2, 2\] M⊙ BNS bracket.** A total mass of 4 M⊙ lies above commonly quoted prompt-collapse thresholds (EOS-dependent, typically ≈ 2.8–3.5 M⊙), a regime the paper itself associates with limited ejecta (Sect. 2.1.1). It is best read as the upper end of the chirp-mass bracket rather than as a preferred scenario.
3. **Minimum-spin choice.** For fixed masses and an aligned-spin model, the minimum BH spin compatible with $M_{\mathrm{REM}}$ > 0 yields the lowest SNR among EM-bright configurations, hence the smallest `D₉₀`: conservative for exclusion, possibly pessimistic for follow-up prioritization.
4. **Coherent versus incoherent statistics.** The TDR uses an incoherent sum with $2N_{\mathrm{IFO}}$ degrees of freedom, whereas PyGRB combines the detector data coherently using the known sky position. How this alters the null distribution and the effective $\rho _{\mathrm{cut}}$ is absorbed by the empirical calibration but not derived.
5. **Priors.** The alignment of the jet axis with the orbital angular momentum and the inclination ranges inferred from jet breaks (with their acknowledged detection biases) propagate directly into `D₉₀`; for poorly constrained triggers the paper recommends running both inclination priors.
6. **Redshift.** The calculation is in terms of luminosity distance; at ≲ 300 Mpc (z ≲ 0.07) the detector-frame chirp mass is shifted by a factor (1+z) at the few-percent level. The paper does not discuss this; it will matter at third-generation reach.

## 7. Wrap-up and discussion (3 min)

**One-slide summary.**

- *Question:* given an EM or neutrino trigger, how far could the LVK network have seen a CBC associated with it?
- *Method:* EM priors (sky map, ι, EM-bright masses) plus the PSD of the IFOs observing at `t₀`; optimal SNR from one reference computation rescaled with $D_{\mathrm{eff}}$; network SNR; conversion to matched-filter SNR through a non-central χ<sup>2</sup> law with mismatch; distance at which 90% of sources exceed $\rho _{\mathrm{cut}}$.
- *Validation:* against PyGRB 90% exclusion distances for 71 GRBs; median Δ<sub>90</sub> = 0.02, 16–84% range \[−0.14, +0.20\] at $\rho _{\mathrm{MF,cut}}$ = 9.
- *Use:* efficiency curves and `D₉₀` maps within minutes, to prioritize follow-up and to interpret transients (GRB 250225B, 251026B, 211211A, 230307A, AT2025ulz, Einstein Probe fast X-ray transients).
- *Caveat:* an expected-sensitivity estimate, not a search.

**Key equations.**

$$
\begin{gathered}
h = F_{+}h_{+} + F_{\times}h_{\times}\quad\text{detector response} \\
\mathcal{M}  = (m_{1}m_{2})^{(3/5)} / (m_{1}+m_{2})^{(1/5)}\quad\text{chirp mass} \\
(a|b) = 4\Re  \int  \tilde{a} \tilde{b}^{*} / S_{n} df ,  \rho _{\mathrm{opt}} = (h|h)^{(1/2)}\quad\text{inner product, optimal SNR} \\
\rho _{\mathrm{MF}}^{2} \sim \text{non-central} \chi ^{2} (2N_{\mathrm{IFO}}, k\rho _{\mathrm{net}}^{2})\quad\text{matched-filter SNR} \\
D_{\mathrm{eff}} = D_{\mathrm{L}}/\Omega  ,  \rho  \propto  \mathcal{M} ^{(5/6)} / D_{\mathrm{eff}}\quad\text{geometry rescales the SNR} \\
P(\rho _{\mathrm{net}} > \rho _{\mathrm{cut}}) = \lambda   \to   D_{90}^{\mathrm{TDR}}\quad\text{definition of the TDR}
\end{gathered}
$$

**Exercises** (answers in brackets).

1. Show that $F_{+}^{2} + F_{\times}^{2}$ is independent of ψ; locate its maximum and its zeros. \[Maximum 1 at θ = 0; zeros at θ = π/2, φ = π/4 + nπ/2.\]
2. Compute ℳ for \[1.4, 10\] M⊙ and the inspiral-only reach relative to a 1.4 + 1.4 BNS. \[ℳ = 3.00 M⊙; (3.00/1.22)^(5/6) ≈ 2.1 before $f_{\mathrm{ISCO}}$ truncation.\]
3. A GRB has $D_{90}^{\mathrm{TDR}}$ = 120 Mpc for BNS. What follows for a candidate host at 250 Mpc? At 60 Mpc? \[At 250 Mpc a GW non-detection is not constraining and the merger remains viable; at 60 Mpc ≪ `D₉₀` the absence of a candidate disfavors a merger, given the priors.\]
4. Why is the best-matching threshold $\rho _{\mathrm{opt}}$ = 10 rather than 9? \[Noise scatter of ≈ 1 and mismatch k raise the optimal SNR needed for 90% efficiency.\]
5. How does the TDR change if one IFO is offline, or if the localization map is 100 times larger? \[Lower network SNR and different sky coverage; a larger map samples less favorable antenna-pattern regions and tends to lower `D₉₀`, because 90% efficiency is limited by the least favorable parts of the map.\]
6. How would you handle a merger time uncertain by ±T? \[Evaluate on successive segments and report the most conservative (smallest) `D₉₀`, as in Sect. 5 of the paper.\]
